# Phase 7 — Inference Engine and Recommendations

This notebook demonstrates `ml/predict.py` for single and batch predictions with recommendation lookup.

```python
import sys
from pathlib import Path
from PIL import Image

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from ml.predict import WastePredictor
```

```python
predictor = WastePredictor(model_name='custom_cnn', confidence_threshold=0.60, top_k=3)
predictor
```

```python
demo_image = Image.new('RGB', (320, 240), color=(140, 180, 220))
result = predictor.predict(demo_image)
result
```

```python
batch_results = predictor.predict_batch([demo_image, demo_image], top_k=2)
batch_results
```
