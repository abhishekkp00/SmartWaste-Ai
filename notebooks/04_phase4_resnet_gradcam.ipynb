# SmartWaste AI — Phase 4: ResNet18 and Grad-CAM

This notebook validates ResNet18 Stage A and Stage B architectures plus Grad-CAM. It does not train the model; training belongs to Phase 5.


```python
from pathlib import Path
import os, sys

PROJECT_ROOT = Path('/content/SmartWaste_Ai') if Path('/content/SmartWaste_Ai').exists() else Path.cwd()
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))
print('Project root:', PROJECT_ROOT)
```

```python
import torch
from ml.resnet_model import build_resnet18_stage_a, build_resnet18_stage_b, count_parameters

stage_a = build_resnet18_stage_a(pretrained=False)
stage_b = build_resnet18_stage_b(pretrained=False)

print('Stage A trainable / total:', count_parameters(stage_a))
print('Stage B trainable / total:', count_parameters(stage_b))
```

```python
from ml.dataset import create_dataloaders
from ml.gradcam import GradCAM, save_heatmap

train_loader, _, _ = create_dataloaders(batch_size=4, num_workers=0)
images, labels = next(iter(train_loader))

stage_a.eval()
with torch.no_grad():
    logits = stage_a(images)
print('Logits:', tuple(logits.shape))

with GradCAM(stage_a) as cam:
    heatmap, predicted = cam.generate(images[:1])
print('Predicted class index:', predicted)
print('Heatmap shape:', heatmap.shape)
save_heatmap(heatmap, PROJECT_ROOT / 'results/eda/phase4_notebook_gradcam.png')
```
