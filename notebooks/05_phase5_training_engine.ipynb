# SmartWaste AI — Phase 5: Training Engine

This notebook demonstrates the reusable training engine, weighted loss, optional sampler, metrics, early stopping, scheduler, checkpointing, and history. Full multi-epoch experiments belong to later execution scripts and Phase 6 evaluation.


```python
from pathlib import Path
import os, sys

PROJECT_ROOT = Path('/content/SmartWaste_Ai') if Path('/content/SmartWaste_Ai').exists() else Path.cwd()
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

from ml.config import cfg
from ml.custom_cnn import build_custom_cnn
from ml.train import create_training_components
```

```python
trainer, train_loader, val_loader, test_loader, weights = create_training_components(
    model=build_custom_cnn(),
    batch_size=4,
    num_workers=0,
    use_weighted_sampler=False,
    weighted_loss=True,
)
print('Class weights:', weights.tolist())
print('Train batches:', len(train_loader))
print('Val batches:', len(val_loader))
print('Test batches:', len(test_loader))
```

```python
train_metrics = trainer.train_epoch(train_loader)
val_metrics = trainer.validate_epoch(val_loader)
print('Train metrics:', train_metrics)
print('Val metrics:', val_metrics)
```
