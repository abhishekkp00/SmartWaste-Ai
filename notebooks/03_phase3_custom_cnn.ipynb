# SmartWaste AI — Phase 3: Custom CNN from Scratch

This notebook validates the five-block Custom CNN. It does **not** train the model or implement ResNet18; training is a later phase.


```python
from pathlib import Path
import os

PROJECT_ROOT = Path('/content/SmartWaste_Ai') if Path('/content/SmartWaste_Ai').exists() else Path.cwd()
if PROJECT_ROOT.name != 'SmartWaste_Ai':
    PROJECT_ROOT = PROJECT_ROOT.resolve()
os.chdir(PROJECT_ROOT)
import sys
sys.path.insert(0, str(PROJECT_ROOT))
print('Project root:', PROJECT_ROOT)
```

```python
import torch
from ml.config import cfg
from ml.custom_cnn import build_custom_cnn, count_parameters

model = build_custom_cnn()
trainable, total = count_parameters(model)

print(model)
print(f'Trainable parameters: {trainable:,}')
print(f'Total parameters: {total:,}')
```

```python
from ml.dataset import create_dataloaders

train_loader, val_loader, test_loader = create_dataloaders(batch_size=8, num_workers=0)
images, labels = next(iter(train_loader))

model.eval()
with torch.no_grad():
    features = model.forward_features(images)
    logits = model(images)

print('Image batch:', tuple(images.shape))
print('Label batch:', tuple(labels.shape), labels.dtype)
print('Feature batch:', tuple(features.shape))
print('Logit batch:', tuple(logits.shape))
```
