# SmartWaste AI — Phase 6: Evaluation and Comparison

This notebook demonstrates test-set evaluation, confusion matrices, classification reports, training-history plots, and model-comparison plots. It evaluates an existing model and does not train or compare real checkpoints automatically.


```python
from pathlib import Path
import os, sys

PROJECT_ROOT = Path('/content/SmartWaste_Ai') if Path('/content/SmartWaste_Ai').exists() else Path.cwd()
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

from ml.custom_cnn import build_custom_cnn
from ml.dataset import create_dataloaders
from ml.evaluate import evaluate_model, plot_confusion_matrix, save_classification_report
```

```python
import torch

_, _, test_loader = create_dataloaders(batch_size=32, num_workers=0)
model = build_custom_cnn()
metrics = evaluate_model(model, test_loader, criterion=torch.nn.CrossEntropyLoss(), device=torch.device('cpu'))

print('Accuracy:', metrics['accuracy'])
print('Macro precision:', metrics['precision_macro'])
print('Macro recall:', metrics['recall_macro'])
print('Macro F1:', metrics['f1_macro'])

plot_confusion_matrix(metrics['targets'], metrics['predictions'])
save_classification_report(metrics)
```
